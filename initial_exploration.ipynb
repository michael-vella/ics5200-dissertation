# Notebook Purpose

The purpose of this notebook is to explore and start getting familiar with the packages and representations used in the drug discovery domain.

## Imports

Import any packages required.

```python
import pandas as pd
from rdkit import Chem
from IPython.display import display
```

## Attributes

Define any attributes required for this notebook.

```python
data_folder_dir = "data"
raw_data_folder_dir = f"{data_folder_dir}/raw"
muv_csv_file_path = f"{raw_data_folder_dir}/muv.csv"
tox21_csv_file_path = f"{raw_data_folder_dir}/tox21.csv"
```

## Methods

Methods used in this notebook.

### Explore Pandas DataFrame

Method to explore pandas DataFrame and display short overview.

```python
def explore_df(df: pd.DataFrame) -> None:
    print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns\n")
    
    print("Info:")
    df.info()
    
    print("\nDescribe:")
    display(df.describe(include="all"))
    
    print("\nSample:")
    display(df.head())
```

### Draw molecules

Draw molecules from smiles string in pandas DataFrame.

```python
def draw_molecules(df: pd.DataFrame, smiles_col: str):
    smile_strings = [row[smiles_col] for _, row in df.iterrows()][:6]
    molecules = [Chem.MolFromSmiles(smile_string) for smile_string in smile_strings]
    print("Smile strings:", smile_strings)
    return Chem.Draw.MolsToGridImage(molecules, molsPerRow=3, subImgSize=(400, 400))
```

## MUV Dataset

Initial exploration of the MUV dataset.

```python
muv_df = pd.read_csv(muv_csv_file_path)

explore_df(muv_df)
```

```python
draw_molecules(muv_df, "smiles")
```

## Tox21 Dataset

Initial exploration of the Tox21 dataset.

```python
tox_21_df = pd.read_csv(tox21_csv_file_path)

explore_df(tox_21_df)
```

```python
draw_molecules(tox_21_df, "smiles")
```
