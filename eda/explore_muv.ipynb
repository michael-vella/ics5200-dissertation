# Notebook Purpose

The purpose of this Jupyter notebook is to explore the MUV dataset. We explore both the raw (CSV) dataset as well as the processed (PKL) datasets.

## Imports

Import any modules requires.

```python
import sys
from pathlib import Path

# make the project root importable so the `datasets` package resolves
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from eda.helpers import *
from datasets.dataset_handler import DatasetHandler
from datasets.enums import DatasetSource, FeatureType
```

## Attributes

```python
handler = DatasetHandler()
dataset_source = DatasetSource.MUV
```

## Raw CSV

Explore the raw CSV file after loading it into a Pandas DataFrame.

```python
raw_df = handler.load_raw_dataset(
    dataset_source=dataset_source
)
```

```python
explore_df(df=raw_df)
```

```python
check_nulls(df=raw_df, column="smiles")
```

## Processed PKL files

Display and look at the processed PKL files created.

### ECFP Fingerprint

```python
ecfp_df = handler.load_processed_dataset(dataset_source=dataset_source, feature_type=FeatureType.ECFP)
show_df(df=ecfp_df)
show_first_value(df=ecfp_df, column="mol")
```

### DGL

```python
dgl_df = handler.load_processed_dataset(dataset_source=dataset_source, feature_type=FeatureType.DGL)
show_df(df=dgl_df)
show_first_value(df=dgl_df, column="mol")
```

### DGL with bonds

```python
dgl_bonds_df = handler.load_processed_dataset(dataset_source=dataset_source, feature_type=FeatureType.DGL_WITH_BONDS)
show_df(df=dgl_bonds_df)
show_first_value(df=dgl_bonds_df, column="mol")
```
