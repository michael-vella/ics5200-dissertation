# Notebook Purpose

EDA on how to generate training/testing tasks (from the experimental assays) given the different feature types.

We experiment with the MUV dataset but same would apply with the other datasets.

## Imports

Import any packages required.

```python
import sys
from pathlib import Path

# make the project root importable so the `datasets` package resolves
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from eda.helpers import *
from datasets.dataset_handler import DatasetHandler
from datasets.enums import DatasetSource, FeatureType
```

## Attributes

Attributes required. As explained in the notebook header, we will attempt this with the MUV dataset for all the different feature types.

This will allow us to understand how test & train tasks are created (depending on the feature types) and will allow us to create a generic process that can be used to create these (agnostic of the dataset source).

```python
test_assays_map = {
    DatasetSource.TOX21: ["SR-HSE", "SR-MMP", "SR-p53"],
    DatasetSource.MUV: ["MUV-832", "MUV-846", "MUV-852", "MUV-858", "MUV-859"],
}

handler = DatasetHandler()
dataset_source = DatasetSource.MUV
```

## Core Logic

Core logic to generate train/test sets - dependence on the feature type chosen.

### ECFP Fingerprint

```python
ecfp_df = handler.load_processed_dataset(dataset_source=dataset_source, feature_type=FeatureType.ECFP)
show_df(ecfp_df)
```

Get the testing tasks (assays) depending on the dataset.

```python
test_assays = test_assays_map.get(dataset_source)
print("Test assays for MUV dataset source:", test_assays)
```

Get the training tasks (assays) by getting the rest of assays that are not inside the testing set.

```python
columns_to_drop = test_assays.copy()
columns_to_drop.extend(["mol_id", "smiles", "mol"])
print("Columns to drop:", columns_to_drop)

train_assays = [x for x in list(ecfp_df.columns) if x not in columns_to_drop]
print("Training assays for MUV dataset:", train_assays)
```

For each assay in the training set, filter out NULLs (NULLs mean that there was no test against that assay for that particular molecule).

```python
train_assays_dict = dict.fromkeys(train_assays)
print("Training assays dictionary:", train_assays_dict)

for assay in train_assays:
    task_df = ecfp_df[[assay, 'mol']].dropna()
    task_df.columns = ['y', 'mol'] # y is the target and mol is the molecular embedding (in this case ECFP)

    # show_df(task_df)

    train_assays_dict[assay] = task_df

print("\nPrinting the length of the DataFrame for each assay (in the training set)...")
for assay, assay_df in train_assays_dict.items():
    print(f"Assay: '{assay}' - DataFrame size: '{len(assay_df)}'")

print("\nPrinting active/inactive counts for each assay (in the training set)...")
for assay, assay_df in train_assays_dict.items():
    assay_value_counts = assay_df["y"].value_counts()
    active_count = assay_value_counts.get(1.0, 0)
    inactive_count = assay_value_counts.get(0.0, 0)

    print(f"Assay: '{assay}'; Active: '{active_count}'; Inactive: '{inactive_count}'")
```

And do the same for the assays in the test set.

```python
test_assays_dict = dict.fromkeys(test_assays)
print("Testing assays dictionary:", test_assays_dict)

for assay in test_assays:
    task_df = ecfp_df[[assay, 'mol']].dropna()
    task_df.columns = ['y', 'mol'] # y is the target and mol is the molecular embedding (in this case ECFP)

    # show_df(task_df)

    test_assays_dict[assay] = task_df

print("\nPrinting the length of the DataFrame for each assay (in the testing set)...")
for assay, assay_df in test_assays_dict.items():
    print(f"Assay: '{assay}' - DataFrame size: '{len(assay_df)}'")

print("\nPrinting active/inactive counts for each assay (in the testing set)...")
for assay, assay_df in test_assays_dict.items():
    assay_value_counts = assay_df["y"].value_counts()
    active_count = assay_value_counts.get(1.0, 0)
    inactive_count = assay_value_counts.get(0.0, 0)

    print(f"Assay: '{assay}'; Active: '{active_count}'; Inactive: '{inactive_count}'")
```

### DGL (Bonds)

```python
dgl_bonds_df = handler.load_processed_dataset(dataset_source=dataset_source, feature_type=FeatureType.DGL_WITH_BONDS)
show_df(dgl_bonds_df)
```

Get the testing tasks (assays) depending on the dataset.

```python
test_assays = test_assays_map.get(dataset_source)
print("Test assays for MUV dataset source:", test_assays)
```
